# Convnet

En este conjunto de celdas se reliza una comparativa entre un modelo Denso y un modelo Convnet para la clasificacion de imagens de digitos del conjunto de datos MNIST.

Primero hay que obtener y procesar los datos de MNIST:

```python
from keras.datasets import mnist
from keras.utils import to_categorical
(train_images, train_labels), (test_images, test_labels) = mnist.load_data()
```

```python
# Para dense
dense_train_images = train_images.reshape((60000, 28 * 28))
dense_train_images = dense_train_images.astype('float32') / 255
dense_test_images = test_images.reshape((10000, 28 * 28))
dense_test_images = dense_test_images.astype('float32') / 255
dense_train_labels = to_categorical(train_labels)
dense_test_labels = to_categorical(test_labels)
```

```python
# Para convnet
conv_train_images = train_images.reshape((60000, 28, 28, 1))
conv_train_images = conv_train_images.astype('float32') / 255
conv_test_images = test_images.reshape((10000, 28, 28, 1))
conv_test_images = conv_test_images.astype('float32') / 255
conv_train_labels = to_categorical(train_labels)
conv_test_labels = to_categorical(test_labels)
```

Modelo base:

```python
from tensorflow.keras import layers
from tensorflow.keras import models

model = models.Sequential()
model.add(layers.Dense(512, activation='relu', input_shape=(28 * 28,)))
model.add(layers.Dense(10, activation='softmax'))

model.summary()
```

```python
model.compile(optimizer='rmsprop', loss='categorical_crossentropy', metrics=['acc'])

base_history = model.fit(
    dense_train_images,
    dense_train_labels,
    epochs=5,
    batch_size=512)

test_loss, test_acc = model.evaluate(dense_test_images, dense_test_labels)
print('test_acc:', test_acc)
```

Acc en test de $0.97$.

Modelo Convnet:

```python
from tensorflow.keras import layers
from tensorflow.keras import models

model = models.Sequential()
model.add(layers.Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1)))
model.add(layers.MaxPooling2D((2,2)))
model.add(layers.Conv2D(64, (3, 3), activation='relu'))
model.add(layers.MaxPooling2D((2, 2)))
model.add(layers.Conv2D(64, (3, 3), activation='relu'))
model.add(layers.Flatten())
model.add(layers.Dense(64, activation='relu'))
model.add(layers.Dense(10, activation='softmax'))

model.summary()
```

```python
model.compile(optimizer='rmsprop', loss='categorical_crossentropy', metrics=['acc'])

model.fit(conv_train_images, conv_train_labels, epochs=5, batch_size=64)

test_loss, test_acc = model.evaluate(conv_test_images, conv_test_labels)
print('test_acc:', test_acc)
```

Acc en Conv = $0.99$

El modelo convolucional es mas ligero que el denso. Se obtiene mayor precicio con un modelo 5 veces mas ligero.

Ahora Quito las capas *MaxPooling2D((2,2))* que reducen la resolución:

```python
from tensorflow.keras import layers
from tensorflow.keras import models

model = models.Sequential()
model.add(layers.Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1)))
model.add(layers.Conv2D(64, (3, 3), activation='relu'))
model.add(layers.Conv2D(64, (3, 3), activation='relu'))
model.add(layers.Flatten())
model.add(layers.Dense(64, activation='relu'))
model.add(layers.Dense(10, activation='softmax'))

model.summary()
```

```python
model.compile(optimizer='rmsprop', loss='categorical_crossentropy', metrics=['acc'])

model.fit(conv_train_images, conv_train_labels, epochs=5, batch_size=64)

test_loss, test_acc = model.evaluate(conv_test_images, conv_test_labels)
print('test_acc:', test_acc)
```

Acc en Conv_no_max_pool = $0.99$

El rendimiento fue similar a la version del modelo con Pool. No obstante el modelo tardo mucho **mas** en entrenar y pesa 6 veces **mas** que la version densa a diferencia de su competidor que pesava 5 veces **menos**.

El *MaxPooling2D* es necesario, no solo reduce el peso y por ende el tiempo de entrenamiento sino que el reducir la resolución permite que los kernels que se aplican sobre capas posteriores a otros puedan encontrar patrones en segmentos mas grandes (respecto a la correspondencia con el input), ya que se aplican sobre pixeles ya filtrados por la capa anterior. Se arma una jerarquía donde cada capa logra identificar patrones mas grandes. La ultima capa convolucional tiene que contener información de la totalidad de la entrada.
