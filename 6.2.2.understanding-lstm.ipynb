# Entendiendo LSTM

*Large Short-Term Memory* (LSTM) fue presentada por Hochreiter y Schmidhuber en 1997 como solucion al problema de *desvanecimiento del gradiente*. Que alega que las redes se vuelven inentrenables en grandes numeros de capas. 

Un ejemplo claro *desvanecimiento* esto ocurre en las redes *feedfoward* donde si las funciones de activacion son de tipo *than* o *sigmoid* el problema hara presencia.

LSTM parte de una red RNN pero añade un camino simultaneo para llevar informacion a lo largo de varios pasos. Lo que se refleja en que se pueden hallar patrones entre elementos mas distantes de una misma serie.

![image.png](images/6.2.2.rnn-to-lstm.png)

El metodo para calcular la salida de un timestep respeta lo visto en *SimpleRNN*:

$$
y = activation(dot(state_t, U) + dot(input_t, W) + b)
$$

Cada una de las 3 transformaciones acarrea sus propias matrices de pesos *i*, *j* y *k*.

```
output_t = activation(dot(state_t, Uo) + dot(input_t, Wo) + dot(C_t, Vo) + bo)
i_t = activation(dot(state_t, Ui) + dot(input_t, Wi) + bi)
f_t = activation(dot(state_t, Uf) + dot(input_t, Wf) + bf)
k_t = activation(dot(state_t, Uk) + dot(input_t, Wk) + bk)
```

Combinando `i_t`, `f_t`, `k_t` el el valor anterior del carril `c_t` se puede calcular el valor proximo del carril `c_t+1`.

```
c_t+1 = i_t * k_t + c_t * f_t
```

La notacion de los indices tiene que ver con los nuevas conexiones que se incorporan a la red:
- Qué entra al estado (input gate → i)
- Qué olvidamos del estado (forget gate → f)
- Qué nuevo candidato a estado considerar (candidate → k)
- Qué sacamos como salida (output gate → o)
